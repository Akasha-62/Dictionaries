```python
# Implement a simple inventory system using dictionaries.
inventory = {}
inventory['cherry'] = inventory.get('cherry', 0) + 3
inventory['grapefruit'] = inventory.get('grapefruit', 0) + 2
inventory['grapefruit'] = max(0, inventory['grapefruit'] - 2)
print(inventory)
```

```python
# Analyze a dataset of customer orders to find the most popular product category.
orders = [
    {'category': 'Solar'},
    {'category': 'Food'}, 
    {'category': 'Solar'},
    {'category': 'Electronics'}
]
categories = [item['category'] for item in orders]
popular = max(set(categories), key=categories.count)
print(popular)
```

```python
inventory = {}
def add(name, qty, price):
    inventory[name] = {'qty': qty, 'price': price}
def update(name, change):
    inventory[name]['qty'] += change
def total():
    return sum(item['qty'] * item['price'] for item in inventory.values())
def low_stock():
    return {name: data for name, data in inventory.items() if data['qty'] < 7}
add('Cherry', 11, 2.7)
add('Grapefruit', 3, 0.75)
update('Cherry', -2)
print(inventory)
print(total())
print(low_stock())
```

# (g)

```python
grades = {
    "Alex": {"Computer": 75, "Biology": 80},
    "Ben": {"Computer": 73, "Biology": 72},
    "Chris": {"Computer": 62, "Biology": 78}
}
averages = {}
for name in grades:
    scores = grades[name].values()
    averages[name] = sum(scores) / len(scores)
computer_scores = [grades[name]["Computer"] for name in grades]
best_computer = max(computer_scores)
worst_computer = min(computer_scores)
sorted_students = sorted(averages.items(), key=lambda x: x[1], reverse=True)
print("Student averages:", averages)
print("Best Computer score:", best_computer)
print("Worst Computer score:", worst_computer)
print("Class ranking:", sorted_students)
```
